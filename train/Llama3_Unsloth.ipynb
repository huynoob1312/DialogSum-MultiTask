# Set up

```python
from unsloth import FastLanguageModel
```

```python
max_seq_length = 2048
dtype = None
load_in_4bit = True
```

```python
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Llama-3.2-3B-Instruct-bnb-4bit",
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)
```

```python
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj",],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",

    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
    use_rslora = False,
    loftq_config = None,
)
model.print_trainable_parameters()
```

```python
def llama_prompt_format(example):
    dialogue = example['dialogue']
    summary = example['summary']
    topic = example['topic']

    return {
        "text": f"""<|begin_of_text|>
<|start_header_id|>system<|end_header_id|>
You are a helpful AI assistant to read the following dialogue and provide a short topic and a brief summary
<|eot_id|>

<|start_header_id|>user<|end_header_id|>
Dialogue: {dialogue}
<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>
Topic: {topic} | Summary: {summary}
<|eot_id|>"""
    }
```

```python
train_dataset = dataset['train']
train_dataset = train_dataset.map(llama_prompt_format)
```

```python
print(llama_prompt_format(dataset['train'][0])['text'])
```

```python
import inspect
from trl import SFTConfig

sft_init_orig = SFTConfig.__init__

def patched_sft_init(self, *args, **kwargs):
    kwargs.pop("push_to_hub_token", None)
    return sft_init_orig(self, *args, **kwargs)

SFTConfig.__init__ = patched_sft_init
```

```python
from trl import SFTTrainer, SFTConfig
import torch

sft_config = SFTConfig(
    dataset_text_field="text",
    max_length=max_seq_length,
    dataset_num_proc=2,
    packing=False,

    per_device_train_batch_size=4,
    gradient_accumulation_steps=4,
    warmup_steps=5,
    num_train_epochs=1,
    learning_rate=2e-4,
    fp16=not torch.cuda.is_bf16_supported(),
    bf16=torch.cuda.is_bf16_supported(),
    logging_steps= 10,
    weight_decay=0.01,
    seed=3407,
    output_dir="outputs",
    report_to="none",

    dataloader_num_workers = 2, 
    dataloader_pin_memory = True, 
)

trainer = SFTTrainer(
    model=model,
    train_dataset=train_dataset,
    args=sft_config,
    processing_class=tokenizer,
    packing = False,
)
```

```python
from unsloth.chat_templates import train_on_responses_only

trainer = train_on_responses_only(
    trainer,
    instruction_part="<|start_header_id|>user<|end_header_id|>\n",
    response_part="<|start_header_id|>assistant<|end_header_id|>\n"
)
```

```python
trainer_stat = trainer.train()
```

# Evaluate

```python
from unsloth import FastLanguageModel
```

```python
max_seq_length = 2048
dtype = None
load_in_4bit = True
```

```python
repo_id = "huynoob1312/dialogsum-multi-task-Llama3.2_3B"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = repo_id,
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)

FastLanguageModel.for_inference(model)
```

```python
tokenizer.padding_side = "left"
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
```

```python
def format_prompt_test(example):
    dialogue = example['dialogue']

    return {
        "text": f"""<|begin_of_text|>
<|start_header_id|>system<|end_header_id|>
You are a helpful AI assistant to read the following dialogue and provide a short topic and a brief summary
<|eot_id|>

<|start_header_id|>user<|end_header_id|>
Dialogue: {dialogue}
<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>
"""
    }
```

```python
test_dataset = dataset["test"].map(format_prompt_test)
```

```python
test_dataset[0]
```

```python
import torch
import re
from tqdm import tqdm

device = torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")

BATCH_SIZE = 16
generated_texts = []

for i in tqdm(range(0, len(test_dataset), BATCH_SIZE)):
    batch_prompts = test_dataset['text'][i : i + BATCH_SIZE]
    
    inputs = tokenizer(
        batch_prompts,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=max_seq_length
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=128,
            use_cache=True,
            do_sample=False
        )
    
    input_lens = inputs.input_ids.shape[1]
    decoded_batch = tokenizer.batch_decode(outputs[:, input_lens:], skip_special_tokens=True)
    generated_texts.extend(decoded_batch)

pred_summaries, ref_summaries = [], []
pred_topics, ref_topics = [], []

for gen_text, ref_sum, ref_top in zip(generated_texts, test_dataset["summary"], test_dataset["topic"]):
    gen_text = gen_text.strip()
    
    if " | Summary: " in gen_text:
        parts = gen_text.split(" | Summary: ")
        pred_topic = re.sub(r'^(Topic[:\s]*)+', '', parts[0], flags=re.IGNORECASE).strip()
        pred_summary = re.sub(r'^(Summary[:\s]*)+', '', parts[1], flags=re.IGNORECASE).strip()
    else:
        pred_topic = "N/A"
        pred_summary = gen_text
    
    pred_summaries.append(pred_summary)
    ref_summaries.append(ref_sum)
    pred_topics.append(pred_topic)
    ref_topics.append(ref_top)
```

```python
import evaluate
rouge = evaluate.load("rouge")
bleu = evaluate.load("sacrebleu")

sum_rouge = rouge.compute(predictions=pred_summaries, references=ref_summaries)
sum_bleu = bleu.compute(predictions=pred_summaries, references=[[r] for r in ref_summaries])

top_rouge = rouge.compute(predictions=pred_topics, references=ref_topics)
top_bleu = bleu.compute(predictions=pred_topics, references=[[r] for r in ref_topics])

print(f"SUMMARY")
print(f"ROUGE-1 : {sum_rouge['rouge1']*100:.2f}%")
print(f"ROUGE-2 : {sum_rouge['rouge2']*100:.2f}%")
print(f"ROUGE-L : {sum_rouge['rougeL']*100:.2f}%")
print(f"BLEU    : {sum_bleu['score']:.2f}")
print("-"*50)
print(f"TOPIC")
print(f"ROUGE-L : {top_rouge['rougeL']*100:.2f}%")
print(f"BLEU    : {top_bleu['score']:.2f}")
```

```python
def format_prompt_predict(dialogue):
    return {
        "text": f"""<|begin_of_text|>
<|start_header_id|>system<|end_header_id|>
You are a helpful AI assistant to read the following dialogue and provide a short topic and a brief summary
<|eot_id|>

<|start_header_id|>user<|end_header_id|>
Dialogue: {dialogue}
<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>
"""
    }

def predict_llama(dialogue: str, max_new_tokens: int = 128):
    prompt = format_prompt_predict(dialogue)['text']
    inputs = tokenizer([prompt], return_tensors="pt").to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            use_cache=True,
            temperature=0.1,    
            do_sample=False     
        )
    
    prompt_len = inputs.input_ids.shape[1]
    response = tokenizer.decode(outputs[0][prompt_len:], skip_special_tokens=True).strip()

    topic, summary = "N/A", response
    if " | Summary: " in response:
        parts = response.split(" | Summary: ")
        topic = re.sub(r'^(Topic[:\s]*)+', '', parts[0], flags=re.IGNORECASE).strip()
        summary = re.sub(r'^(Summary[:\s]*)+', '', parts[1], flags=re.IGNORECASE).strip()

    return {
        "topic": topic,
        "summary": summary,
        "raw_response": response
    }
```
